```python
import networkx as nx
import matplotlib.pyplot as plt

# 1. Define the relative path from src/main.ipynb to the dataset
data_path = "../dataset/com-dblp.ungraph.txt"

# 2. Load the graph
# comments='#' ignores the file headers automatically
# nodetype=int ensures node IDs are treated as numbers, not strings
print("Loading graph into memory...")
G = nx.read_edgelist(data_path, comments='#', nodetype=int, create_using=nx.Graph())

print(f"Original Graph Loaded -> Nodes: {G.number_of_nodes():,}, Edges: {G.number_of_edges():,}")
```
